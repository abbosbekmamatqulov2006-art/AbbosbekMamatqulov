# Data Transformation Plan (Role: Data Transformation Engineer)

## 1. Data Type Transformation
- Raqamli ustunlarni `pd.to_numeric(errors='coerce')`
- Sana ustunlarini `pd.to_datetime(format='%Y-%m-%d')`
- Stringlarni `str.strip()`, `str.title()`
- Kategoriyalarni `astype('category')`

## 2. Missing Value Handling (No Row Deletion)
- **Integer/Float:** o‘rtacha qiymat (mean) bilan to‘ldirish
- **Age:** o‘rtacha yosh bilan to‘ldirib yaxlitlash
- **Date:** `ffill` (oldingi sana bilan)
- **String:** `"Unknown"` bilan to‘ldirish

## 3. Consistency & Standardization
- Sana formatini yagona `YYYY-MM-DD` qilish
- Ism, gender, boolean qiymatlarni standartlashtirish
- Ustun nomlarini kichik harflar va `_` bilan formatlash

## 4. Outlier Treatment
- IQR bilan aniqlab, **winsorization** orqali chegaralash (olib tashlanmaydi)

## 5. Validation & Sorting
- Null qiymatlar qolmaganini tekshirish
- Data type va formatlarni qayta ko‘rib chiqish
- Sana yoki ID bo‘yicha tartiblash

## 6. Export Clean Data
- `df.to_csv("cleaned_dataset.csv", index=False, encoding="utf-8")`
- Keyingi bosqichga (analyst/vizualization) topshirish

```python
df.drop_duplicates(inplace=True)
df
```

```python
# Ustun nomlarini formatlash
df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_").str.replace("[^a-z0-9_]", "", regex=True))
df
```

```python
# barcha string qiymatlarni bir xil formatga keltirish
for col in df.select_dtypes(include='object'):
    df[col] = df[col].astype(str).str.strip().str.title()
df
```

```python
datetime_cols = ['hire_date']
for col in datetime_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
category_cols = ['gender', 'department_id', 'department', 'category',
    'location_id', 'country', 'work_mode',
    'manager_id', 'manager_name']
for col in category_cols:
    df[col] = df[col].astype('category')
print(df.dtypes)
```

```python
# age ustunini integerga o'zgartirib, null qiymatlarni o'rtachasi bilan to'ldirdik
df['age'] = df['age'].fillna(df['age'].mean()).round().astype(int)
df
```

```python
# yoshi manfiy va 110 dan kattalarni aniqlab o'rtacha bilan to'ldirish

mean_age = df.loc[(df['age'] >= 0) & (df['age'] <= 110), 'age'].mean()
df.loc[(df['age'] < 0) | (df['age'] > 110), 'age'] = int(mean_age)
df['age'] = df['age'].astype(int)
df
```

```python
df['hire_date'] = df['hire_date'].ffill()
df.head()
```

```python
df['country'] = df['country'].replace({'Uae': 'United States'})
df
```

```python
# manager idni floatga almashtirib oldik chunki category ustun ustida median qiymatni topish va to'ldirishni amalga oshirib bo'lmaydi
df['manager_id'] = df['manager_id'].astype(float)
median_manager = df['manager_id'].median()
df['manager_id'] = df['manager_id'].fillna(median_manager)

# 2. Int tipiga o‘tkazish
df['manager_id'] = df['manager_id'].astype(int)
df
```
