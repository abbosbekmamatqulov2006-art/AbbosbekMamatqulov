````python
df["Symbol"].value_counts().plot(kind="line")
plt.title("Category Counts")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
````

````python
sns.countplot(x="Symbol", data=df)
plt.title("Category Counts")
plt.show()
````

````python
plt.plot(df["High"], df["Low"])
plt.title("Line Chart Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
````

````python
sns.scatterplot(x="Open", y="Close", data=df, hue=df["Open"].mean())
plt.title("Scatter Plot Example")
plt.xlabel("Column 1")
plt.title("Scatter Plot with mean")
plt.show()
````
